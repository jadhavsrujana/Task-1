```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/bank-full.csv', sep=';')

print(df.head())
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
import matplotlib.pyplot as plt

# Create the figure
plt.figure(figsize=(10, 6))

# Create the histogram
plt.hist(
    df['age'],
    bins=15,
    edgecolor='black'
)

# Add title and labels
plt.title('Distribution of Bank Customers by Age')
plt.xlabel('Age')
plt.ylabel('Number of Customers')

# Add gridlines
plt.grid(axis='y', linestyle='--', alpha=0.4)

# Adjust the layout
plt.tight_layout()

# Display the chart
plt.show()
```
